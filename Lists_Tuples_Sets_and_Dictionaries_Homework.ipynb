## Exercise 4 · Update the game-box stock (about 10 min)


Each dictionary key is a game name; its value is the number of boxes.
1. Change the Azul quantity to 4.
2. Add Dixit with 2 boxes.
3. Delete Chess from the dictionary.
4. Look up `target`. Print its quantity if it exists, otherwise print a short missing-game message. A missing game must not crash or be added to stock.
5. Loop through the updated dictionary to show each game and quantity, and calculate the total number of boxes.

**Check:** the updated stock contains Uno: 3, Azul: 4 and Dixit: 2. Looking up Uno gives 3. There are 9 boxes altogether.

**Try a change:** set `target = "Twister"`. You should get the missing-game message, and stock should still have the same three games and 9 boxes.


```python
stock = {"Uno": 3, "Azul": 1, "Chess": 2}
target = "Uno"
stock = {"Uno": 3, "Azul": 1, "Chess": 2}
target = "Uno"

stock["Azul"] = 4
stock["Dixit"] = 2

del stock["Chess"]

if target in stock:
    print(f"{target}: {stock[target]} boxes")
else:
    print(f"{target} is missing from stock")

total = 0

for game, quantity in stock.items():
    print(game, quantity)
    total += quantity

print("Total boxes:", total)
```

## Exercise 5 · Process the borrowing requests (about 12 min)


Visit each game in `requests`, in order, using a loop. Keep one `stock` dictionary for the whole run.
- If a game is missing from stock, report `Unknown game` and its name.
- If its quantity is zero, report `Unavailable` and its name.
- Otherwise, subtract one box and report `Borrowed` and its name.

After the loop, print the remaining stock. Do not add unknown games or let a quantity become negative. All starting quantities are whole numbers, zero or above.

**Check:** the supplied requests borrow Uno twice and Azul once. Dixit and the third Uno request are unavailable; Twister is unknown. The final stock is `{"Uno": 0, "Azul": 0, "Dixit": 0}`.

**Try changes:** with `requests = []`, stock is unchanged. With `requests = ["Twister", "Dixit"]`, nothing is borrowed and stock is unchanged. Do not copy the sample responses into print calls: the program must respond to the data.


```python
stock = {"Uno": 2, "Azul": 1, "Dixit": 0}
requests = ["Uno", "Dixit", "Twister", "Uno", "Uno", "Azul"]
stock = {"Uno": 2, "Azul": 1, "Dixit": 0}
requests = ["Uno", "Dixit", "Twister", "Uno", "Uno", "Azul"]
for game in requests:
    if game not in stock:
        print("Unknown game:", game)
    elif stock[game] == 0:
        print("Unavailable:", game)
    else:
        stock[game] -= 1
        print("Borrowed:", game)
print("Remaining stock:", stock)
```

## Optional extension · Which games were actually borrowed?
Extend Exercise 5 in a separate cell. Keep the successful game names in a list, including repeats. At the end, show that list and the different borrowed games. An unavailable or unknown request must not enter the borrowed list. For the sample, the successful list is `["Uno", "Uno", "Azul"]`, and there are two different borrowed games. This extension does not affect your grade.


```python
stock = {"Uno": 2, "Azul": 1, "Dixit": 0}
requests = ["Uno", "Dixit", "Twister", "Uno", "Uno", "Azul"]
borrowed_games = []
for game in requests:
    if game not in stock:
        print("Unknown game:", game)
    elif stock[game] == 0:
        print("Unavailable:", game)
    else:
        stock[game] -= 1
        borrowed_games.append(game)
        print("Borrowed:", game)
print("Borrowed list:", borrowed_games)
print("Different borrowed games:", set(borrowed_games))
```
