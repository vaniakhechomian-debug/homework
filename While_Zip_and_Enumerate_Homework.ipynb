## Quick reference: `while`
A `while` loop checks its condition before each turn. Something inside must change the condition or use `break` to end the loop. Here the changing value is `minutes`.

````python
minutes = 2
while minutes > 0:
    print("Minutes:", minutes)
    minutes -= 1


   
````

## Quick reference: lists and `enumerate`
A list keeps several values in order, inside square brackets. A loop can visit those values without using their positions to look them up.

`enumerate()` gives you two values each turn: a number and the next item. `start=1` starts numbering at 1; without it, numbering starts at 0. Use two variable names before `in`. Run this example, then try a different starting number.

````python
books = ["Dune", "Coraline"]
for number, book in enumerate(books, start=1):
    print(number, book)
````

## Quick reference: `zip`
`zip()` pairs values by their order: first with first, second with second. Here each city is paired with its temperature. Use two variable names to receive the pair.

If one list is shorter, ordinary `zip()` stops when that list ends. It does not fill missing values or report an error. Keep related data in the same order.

````python
cities = ["Vanadzor", "Gyumri"]
temperatures = [8, 11]
for city, temperature in zip(cities, temperatures):
    print(city, temperature)
````

### Exercise 3: Book tickets while seats remain (5 points · about 12 min)
There are 5 tickets available. Use `while` to ask how many tickets the next person wants.

- `0` stops taking bookings.
- A request larger than the number remaining prints `Not enough tickets`. Keep the remaining number unchanged and ask again.
- Otherwise, subtract the request and print `Remaining:` followed by the new number.
- When no tickets remain, stop automatically. Print `Booking closed` once after the loop.

Assume each entry is a whole number, zero or above.

**Checks:** requests `2`, `4`, `3` leave 3 tickets, reject 4, then leave 0 and stop without another question. Request `0` first: only Booking closed. Request `5` first: Remaining: 0, then Booking closed.

````python
remaining = 5

while remaining > 0:
    tickets = int(input("How many tickets do you want? "))
    print("Tickets" , tickets)
    if tickets > remaining:
        print("Not enough Tickets")
        continue
    elif remaining == 0:
            break
    remaining -= tickets
    print("Remaining", remaining)

print("Booking Closed")
````
