```python
print ("Group trip budget calculator")
travelers = int(input("please enter the number of travelers"))
print (travelers , "person")
nights = int(input("how many nights do you want to stay?"))
print (nights , "nights")
trip_days = (nights + 1)
print (trip_days , "days")
nightly_whole_group_accommodation = float(input("how much does 1 night costs per person?"))
print (" 1  night per person costs" , nightly_whole_group_accommodation)
print ("total for each person costs" , nights * nightly_whole_group_accommodation)
print ("total nights for whole group accommodation cost " , travelers * nights * nightly_whole_group_accommodation)
whole_trip_group_accommodation = float(input("how much does transfers costs for per person?"))
print (" transfers costs for each person" ,whole_trip_group_accommodation )
print ("whole trip group accommodation costs ", whole_trip_group_accommodation * travelers)
daily_food = float(input("how much does the daily food costs for per person?"))
print (" food for 1 day per person costs" , daily_food)
print("totall for each person costs" ,daily_food * trip_days )
print ("total food costs" , travelers * daily_food * trip_days)
total = travelers * nights * nightly_whole_group_accommodation + whole_trip_group_accommodation * travelers + daily_food * travelers * trip_days
print (total)
```
