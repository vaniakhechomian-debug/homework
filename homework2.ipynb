```python
print ("Group trip budget calculator")
travelers = int(input("please enter the number of travelers"))
nights = int(input("how many nights do you want to stay?"))
nightly_whole_group_accommodation = travelers * 100000
print(nightly_whole_group_accommodation)
whole_trip_group_accommodation = travelers * 2000
print(whole_trip_group_accommodation)
daily_food = travelers * 6000
print(daily_food)
total = nights * (nightly_whole_group_accommodation + daily_food) + whole_trip_group_accommodation
print(total)
```
